```python
import torch
import time

# 1. HARDWARE CHECK
print("="*40)
print("🔍 HARDWARE CHECK")
print("="*40)

if torch.cuda.is_available():
    print("✅ GPU is Active!")
    print(f"Device Name: {torch.cuda.get_device_name(0)}")
else:
    print("❌ GPU Not Found. The Notebook is running on CPU only.")
    print("Tip: Go to the right menu under 'Accelerator' and activate the GPU T4x2.")

print("\n" + "="*40)
print("⏱️ SPEED TEST (BENCHMARK)")
print("="*40)

# Matrix size (A heavy multiplication to stress the processor)
T = 10000

# 2. CPU TEST (Standard Processor)
print("\nGenerating heavy matrices on CPU...")
cpu_tensor_a = torch.randn(T, T)
cpu_tensor_b = torch.randn(T, T)

print("Starting CPU computation...")
start_time_cpu = time.time()
resultado_cpu = torch.matmul(cpu_tensor_a, cpu_tensor_b)
end_time_cpu = time.time()

tempo_cpu = end_time_cpu - start_time_cpu
print(f"🐌 CPU computation time: {tempo_cpu:.4f} seconds")


# 3. GPU TEST (Graphics Card)
if torch.cuda.is_available():
    print("\nSending heavy matrices to GPU...")
    # Move data to GPU memory
    gpu_tensor_a = cpu_tensor_a.cuda()
    gpu_tensor_b = cpu_tensor_b.cuda()
    
    print("Starting GPU computation...")
    # Quick GPU 'warm-up' (benchmark best practices)
    _ = torch.matmul(gpu_tensor_a, gpu_tensor_b)
    torch.cuda.synchronize()
    
    start_time_gpu = time.time()
    resultado_gpu = torch.matmul(gpu_tensor_a, gpu_tensor_b)
    torch.cuda.synchronize() # Wait for GPU to finish before stopping the clock
    end_time_gpu = time.time()
    
    tempo_gpu = end_time_gpu - start_time_gpu
    print(f"🚀 GPU computation time: {tempo_gpu:.4f} seconds")
    
    # 4. COMPARISON
    print("\n" + "="*40)
    print("🏆 FINAL RESULT")
    print("="*40)
    print(f"The GPU was {tempo_cpu / tempo_gpu:.1f}x times faster than the CPU!")
else:
    print("\n⚠️ GPU test canceled. Please turn on the accelerator in the right menu.")
```
